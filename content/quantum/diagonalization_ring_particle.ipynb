# Electron on a Ring in a Constant Electric Field



An electron of mass $m$ is confined to a circular ring with radius $R$. It is natural to describe it with a 1D wavefunction of the azimuthal angle, $\psi(\phi)$. The wavefunction must satisify the condition $\psi(\phi) = \psi(\phi + 2\pi)$, as well as the normalization condition 

$$ \int_0^{2\pi} |\psi(\phi)|^2 d\phi = 1.$$

Before the electric field is turned on, the (unperturbed) Hamiltonian is simply the kinetic energy operator, which can be written as

$$ \hat{H}^{(0)} = \frac{\hat{L}_z^2}{2mR^2} = -\frac{\hbar^2}{2mR^2} \frac{d^2}{d\phi^2}.$$


### (a)

Starting from the ansatz $\psi = A \exp(in\phi)$, show that when the electric field is zero, the normalized eigenstates of the ring are

$$ \psi_n^{(0)} = \frac{1}{\sqrt{2\pi}} \exp (in\phi), $$

with corresponding eigenvalues 

$$ E_n^{(0)} = \frac{\hbar^2 n^2}{2mR^2}, \quad n \in \mathbb{Z}. $$

Note that with the exception of the ground state, all the eigenstates are doubly degenerate $(E_n = E_{-n})$

### Solution

<details style="display: inline;"><summary style="display: inline; cursor: pointer;">(click to show)</summary><span style="display: block; background: #f9f9f9; padding: 5px; border-left: 3px solid #007acc; margin-top: 5px;"> 

There are three things that we need to show:

1) That the suggested ansatz is an eigenfunction of the unperturbed Hamiltonian with eigenvalue $E_n^{(0)} = \frac{\hbar^2 n^2}{2mR^2}$. 
2) That the periodicity of the system forces $n$ to be an integer. 
3) That the appropriate normalization constant is $ A = 1/\sqrt{2\pi} $

If the suggested ansatz is an eigenfunction, then it satisfies

$$ -\frac{\hbar^2}{2mR^2} \frac{d^2}{d\phi^2} A \exp(in\phi) = E_n A \exp(in\phi), $$

for some constant $E_n$. Taking the second derivative, we find that

$$ -\frac{\hbar^2}{2mR^2} (in)^2 A \exp(in\phi) = E_n A \exp(in \phi),  $$

and hence that

$$ \frac{\hbar^2 n^2}{2mR^2} A \exp(in\phi) = E_n A \exp(in\phi). $$

Therefore, the provided ansatz is indeed an eigenfunction of the Hamiltonian with eigenvalue $E_n = \frac{\hbar^2 n^2}{2mR^2}$. 

Next, we demand that the function have period $2\pi$, i.e., that $\psi(\phi) = \psi(\phi + 2 \pi)$.

$$ A \exp(in\phi) = A \exp(in(\phi+2\pi)) = A \exp(in\phi) \exp(in 2 \pi) $$

$$ 1 = \exp(in 2 \pi) $$

The last equality holds if and only if the variable $n$ is an integer. 

Lastly, we normalize the eigenfunctions:

$$ 1 = \langle \psi_n^{(0)} |\psi_n^{(0)} \rangle = \int_0^{2\pi} A^* \exp (-in\phi) A \exp (in\phi) d\phi = |A|^2 \int_0^{2\pi} d\phi = 2\pi |A|^2 $$

Taking A to be positive and real, we find that $A = 1/\sqrt{2\pi}$ for all of the eigenfunctions, and the proof is complete. 

</span></details></div>

### (b)
<div> A constant electric field $F\hat{x}$ is now applied along the $\hat{x}$-direction, in the plane of the ring, giving rise to a perturbative electric potential term in the Hamiltonian, $H' = eFR \cos (\phi)$, where $e$ is the (positive) elementary charge. Calculate the general matrix element $H_{mn}' = \langle \psi_m^{(0)} |H'|\psi_n^{(0)} \rangle$. 

### Solution 

<details style="display: inline;"><summary style="display: inline; cursor: pointer;">(click to show)</summary><span style="display: block; background: #f9f9f9; padding: 5px; border-left: 3px solid #007acc; margin-top: 5px;"> 

$$H_{mn}' = \langle \psi_m^{(0)} |H'|\psi_n^{(0)} \rangle = \int_0^{2\pi} (\psi_m^{(0)})^* eFR \cos (\phi) \psi_n^{(0)} d \phi = \frac{eFR}{2\pi} \int_0^{2\pi} e^{-im\phi} \cos (\phi) e^{in\phi} d\phi = \frac{eFR}{2\pi}\int_0^{2\pi} e^{i(n-m)\phi} \cos (\phi) d \phi $$

Writing cosine in its complex exponential form, we have that 

$$H_{mn}' = \frac{eFR}{2\pi} \int_0^{2\pi} e^{i(n-m)\phi} \frac{e^{i\phi} + e^{-i]\phi}}{2} d \phi = \frac{eFR}{4\pi} \int_0^{2\pi} e^{i(n-m+1)\phi} + e^{i(n-m-1)\phi} d\phi$$

Noting that $\int_0^{2\pi} e^{ik\phi} d\phi$ is equal to $2\pi$ when $k=0$, and equal to zero when $k$ is any other integer, we obtain the following result:

$$H_{mn}' = \frac{eFR}{2} (\delta_{n, m+1}  + \delta_{n, m-1}) $$

In other words, the matrix element between any two states will vanish unless their indices differ by exactly one, in which case it is equal to $\frac{eFR}{2}$. 

</span></details></div>

### (c)
Calculate the energy of the (non-degenerate) ground state to second order in the electric field strength $F$. 

### Solution 

<details style="display: inline;"><summary style="display: inline; cursor: pointer;">(click to show)</summary><span style="display: block; background: #f9f9f9; padding: 5px; border-left: 3px solid #007acc; margin-top: 5px;"> 

The first order correction to the energy is simply

$$ H_{00}' = \frac{eFR}{2} (\delta_{0, 1}  + \delta_{0, -1}) =0. $$

The second-order correction to the energy is given by 

$$ E_0^{(2)} = \sum_{m\neq 0} \frac{|H_{m0}'|^2}{E_0 - E_m}, $$

but the only terms with non-zero matrix elements are those with $m=\pm$, with corresponding energies $E_{\pm 1} = \frac{\hbar^2}{2mR^2}$. We therefore obtain

$$ E_0^{(2)} = \frac{(\frac{eFR}{2})^2}{0- \frac{\hbar^2}{2mR^2}} + \frac{(\frac{eFR}{2})^2}{0- \frac{\hbar^2}{2mR^2}} = -2\frac{2mR^2}{\hbar^2} (\frac{eFR}{2})^2$$

$$ E_0^{(2)} =  -\frac{m e^2 R^4}{\hbar^2} F^2 $$

</span></details></div>

### (d)
Show that the degeneracy of the $n \neq 0$ states is not broken to first-order. 

### Solution

<details style="display: inline;"><summary style="display: inline; cursor: pointer;">(click to show)</summary><span style="display: block; background: #f9f9f9; padding: 5px; border-left: 3px solid #007acc; margin-top: 5px;"> 

Each pair of states indexed by $\pm n$ form a two-dimensional degenerate subspace. If we knew what the 'good' linear combinations of these states were, we could use them calculate the first-order energy corrections from a single inner product each. However, since we have not yet determined the good states, obtaining the first-order energy corrections requires us to construct first the matrix representation of $H'$ in the degenerate subspace, then diagonalize it. 

$$ H' \to \begin{bmatrix}
H_{n,n}' & H_{n, -n}' \\
H_{-n, n}' & H_{-n, -n}'
\end{bmatrix} $$

But $n$ and $-n$ can never differ by exactly one, and we have already shown that $H_{n,n}'=0$, so the matrix elements are all zero. 

$$ H' \to \begin{bmatrix}
0 & 0 \\
0 & 0
\end{bmatrix} $$

The first-order energy corrections are the eigenvalues of the above matrix. Trivially, these are 0 and 0, so the degeneracy is not broken at first order. However, since this is a constant multiple of the identity matrix, the eigenvectors could still be anything. In this case, determining the good states requires us to go to a higher order of perturbation theory.

</span></details></div>

### (e)

Find the energy of the states $ n = \pm 1$ to second-order in the perturbation. Since the degeneracy of these states was not broken at first order, we weren't able to identify the good states. To do that, we need to go to second-order. Instead of diagonalizing $ W_{kl} = \langle \psi_k^{(0)}| H' | \psi_l^{(0)} \rangle$, diagonalize the matrix

$$ M_{kl} = \sum_{p} \frac{ \langle \psi_{n,k}^{(0)}| H' | \psi_{p}^{(0)} \rangle \langle \psi_{p}^{(0)}| H' | \psi_{n,l}^{(0)} \rangle}{E_n^{(0)}- E_p^{(0)}}, $$

where the two degenerate states within the $n=1$ degenerate subspace are indexed by $k,l$, and the sum over $p$ runs over all states that are not degenerate with the two being corrected (i.e. every other state). 

### Solution

<details style="display: inline;"><summary style="display: inline; cursor: pointer;">(click to show)</summary><span style="display: block; background: #f9f9f9; padding: 5px; border-left: 3px solid #007acc; margin-top: 5px;"> 

$$ 
M_{00} = \sum_{p} \frac{ \langle \psi_{1,0}^{(0)}| H' | \psi_{p}^{(0)} \rangle \langle \psi_{p}^{(0)}| H' | \psi_{1,0}^{(0)} \rangle}{E_1^{(0)}- E_p^{(0)}}
          = \sum_{p} \frac{ \langle \psi_{1}^{(0)}| H' | \psi_{p}^{(0)} \rangle \langle \psi_{p}^{(0)}| H' | \psi_{1}^{(0)} \rangle}{E_1^{(0)}- E_p^{(0)}}
          = \sum_{p} \frac{|H_{1,p}'|^2 }{E_1^{(0)}- E_p^{(0)}}
$$

The matrix element is only non-zero when $p=0$, or when $p=2$. The sum therefore collapses down to 

$$ M_{00}= \frac{(\frac{eFR}{2})^2}{\frac{\hbar^2}{2mR^2}(1^2 - 0^2)} +  \frac{(\frac{eFR}{2})^2}{\frac{\hbar^2}{2mR^2}(1^2 - 2^2)} = \frac{m e^2 R^4}{2\hbar^2} F^2 (1 - \frac{1}{3})$$

$$ M_{00} = \frac{m e^2 R^4}{3\hbar^2} F^2$$

The calculation for $M_{11}$ is essentially identical, except that the non-zero contributions come from the states $p=0$ and $p=-2$. We find that 

$$ M_{11} = \frac{m e^2 R^4}{3\hbar^2} F^2 $$

For the off-diagonal element, we have 

$$
M_{10} = \sum_{p} \frac{ \langle \psi_{1,0}^{(0)}| H' | \psi_{p}^{(0)} \rangle \langle \psi_{p}^{(0)}| H' | \psi_{1,1}^{(0)} \rangle}{E_1^{(0)}- E_p^{(0)}}
        = \sum_{p} \frac{ \langle \psi_{1}^{(0)}| H' | \psi_{p}^{(0)} \rangle \langle \psi_{p}^{(0)}| H' | \psi_{-1}^{(0)} \rangle}{E_1^{(0)}- E_p^{(0)}}
        = \sum_{p} \frac{H_{1,p}' H_{-1, p}'}{E_1^{(0)}- E_p^{(0)}}
$$

Note that while $p=2$ makes the first matrix element non-zero, it causes the second matrix element to vanish. Similarly, $p=-2$ makes the second matrix element non-zero, but causes the first matrix element to vanish. The only value of $p$ for which **both** matrix elements are non-zero is $p=0$, so the sum collapses down to 

$$ M_{10} = \frac{(\frac{eFR}{2})^2}{\frac{\hbar^2}{2mR^2}(1^2 - 0^2)} = \frac{m e^2 R^4}{2\hbar^2} F^2 = M_{01},  $$

where the last equality follows from the fact that the matrix $M$ is Hermitian. 

$$ M = \frac{m e^2 R^4}{\hbar^2} F^2 \begin{bmatrix}
\frac{1}{3} & \frac{1}{2} \\
\frac{1}{2}& \frac{1}{3}
\end{bmatrix} $$

Ignoring the constant factor out front, the eigenvalues of the matrix are easily shown to be $\lambda = \frac{5}{6}, -\frac{1}{6}$. The second-order energy corrections to the $n=\pm 1$ states are therefore 

$$ E_{1, 0}^{(2)} = \frac{5}{6} \frac{m e^2 R^4}{\hbar^2} F^2, \quad  E_{1, 1}^{(2)} = -\frac{1}{6} \frac{m e^2 R^4}{\hbar^2} F^2 $$

The eigenvectors of this matrix give us the good linear combinations of our original basis states. In this case, the eigenvectors are simply $(1, 1)^T$ and $(1, -1)^T$, which means that the good eigenstates are (up to normalization) the familiar functions $\cos (\phi)$ and $\sin(\phi)$. 

</span></details></div>

### (f)

Construct the $M_{kl}$ matrix for the $n$-th degenerate subspace with $n>1$. Show that while there is a non-zero 2nd order correction to the energies, it does not break their degeneracy. 

### Solution

<details style="display: inline;"><summary style="display: inline; cursor: pointer;">(click to show)</summary><span style="display: block; background: #f9f9f9; padding: 5px; border-left: 3px solid #007acc; margin-top: 5px;"> 

The approach is identical to the previous problem, except that for $n>1$, there is no longer any value of $p$ that is simultaneously one away from both $n$ and $-n$. The off-diagonal matrix elements $M_{n, -n}$ are therefore zero in all of the remaining degenerate subspaces. 

For the diagonal elements, $|H_{n, p}'|^2$ is non-zero when $p=n\pm 1$. We find that 

$$ M_{n,n}= \sum_{p} \frac{|H_{n,p}'|^2 }{E_n^{(0)}- E_p^{(0)}} = \frac{(\frac{eFR}{2})^2}{\frac{\hbar^2}{2mR^2}(n^2 - (n-1)^2)} + \frac{(\frac{eFR}{2})^2}{\frac{\hbar^2}{2mR^2}(n^2 - (n+1)^2)} = \frac{m e^2 R^4}{2\hbar^2} F^2 (\frac{1}{2n-1} + \frac{1}{-2n-1}). $$

After some simplification, we find that 

$$ M_{n,n} = \frac{m e^2 R^4}{\hbar^2} F^2 \frac{1}{4n^2-1} = M_{-n, -n},  $$

so the $M$ matrix is 

$$ M = \frac{m e^2 R^4}{\hbar^2} F^2 \begin{bmatrix}
\frac{1}{4n^2-1} & 0 \\
0 & \frac{1}{4n^2-1}
\end{bmatrix} $$

The second-order energy corrections are the eigenvalues of this matrix, which can be read off immediately. We see that while both states receive a non-zero correction at second order, they are both corrected by the same amount, so the degeneracy does not split. However, we cannot determine the good states, since our matrix is a constant multiple of the identity matrix. 

In general one needs to go to $2n$-th order in perturbation theory to see a breaking of the degeneracy of the $n$-th pair of states (for this particular problem).

# Numerical Solutions (Matrix Representation)

We will now introduce a standard technique for obtaining solutions to 1D problems like the one considered above - mapping our continuous differential equation into a matrix equation. 

Our wavefunctions $\psi(\phi)$ are defined on the periodic domain $[0, 2\pi)$. If we represent this domain with a uniform grid of discrete values $\phi_n = n(\Delta \phi) = n \frac{2\pi}{N}, \quad n = 0, 1, 2, ..., N-1$, we can represent our wavefunctions as vectors whose elements are the wavefunction's value at those discrete points:

$$ \psi \to (\psi_0, \psi_1, \psi_2, ..., \psi_{N-1}) = (\psi(\phi_0), \psi(\phi_1), \psi(\phi_2), ..., \psi(\phi_{N-1})) $$

Recall the standard expression for the second-derivative of a function whose value is only known at discrete points:

$$ \frac{d^2\psi}{d\phi^2}|_{\phi_n} \approx \frac{\frac{d\psi}{d\phi}|_{\phi_{n+1}} - \frac{d\psi}{d\phi}|_{\phi_n}}{\phi_{n+1}-\phi_n}
= \frac{\psi_{n+1} - \psi_{n} - (\psi_n - \psi_{n-1})}{(\phi_{n+1}-\phi_n)^2} = \frac{\psi_{n+1} - 2\psi_{n} + \psi_{n-1}}{(\Delta \phi)^2}$$

The vector representation of the second derivative of our wavefunction, $\frac{d^2\psi}{d\phi^2}$, is therefore given by 

$$ \frac{1}{(\Delta \phi)^2}\begin{pmatrix}
-2 & 1 & 0 & \cdots & 0 & 1 \\
1 & -2 & 1 & \ddots & \vdots & \vdots \\
0 & 1 & -2 & \ddots & 0 & 0 \\
\vdots & \ddots & \ddots & \ddots & 1 & 0 \\
0 & \cdots & 0 & 1 & -2 & 1 \\
1 & \cdots & 0 & 0 & 1 & -2
\end{pmatrix}
\begin{pmatrix}
\psi_0 \\
\psi_1 \\
\vdots \\
\psi_{N-2} \\
\psi_{N-1}
\end{pmatrix},
$$

where the elements in the upper-right and lower-left corners are a consequence of our system's periodic boundaries. 

The full Schrodinger equation is then mapped to a discrete problem in the following manner:

$$ 
-\frac{\hbar^2}{2mR^2} \frac{d^2}{d\phi^2} \psi + eFR \cos (\phi) \psi = E \psi \to 
\frac{\hbar^2}{2m(\Delta \phi)^2}
\begin{pmatrix}
2 & -1 & 0 & \cdots & 0 & -1 \\
-1 & 2 & -1 & \ddots & \vdots & \vdots \\
0 & -1 & 2 & \ddots & 0 & 0 \\
\vdots & \ddots & \ddots & \ddots & -1 & 0 \\
0 & \cdots & 0 & -1 & 2 & -1 \\
-1 & \cdots & 0 & 0 & -1 & 2
\end{pmatrix}
\begin{pmatrix}
\psi_0 \\
\psi_1 \\
\vdots \\
\psi_{N-2} \\
\psi_{N-1}
\end{pmatrix}
+eFR \begin{pmatrix}
\cos \phi_0 & 0   & 0   & \cdots & 0   & 0 \\
0   & \cos \phi_1 & 0   & \cdots & 0   & 0 \\
0   & 0   & \cos \phi_2 & \cdots & 0   & 0 \\
\vdots & \vdots & \vdots & \ddots & \vdots & \vdots \\
0   & 0   & 0   & \cdots & \cos \phi_{N-2} & 0 \\
0   & 0   & 0   & \cdots & 0   & \cos \phi_{N-1}
\end{pmatrix}
\begin{pmatrix}
\psi_0 \\
\psi_1 \\
\vdots \\
\psi_{N-2} \\
\psi_{N-1}
\end{pmatrix} = 
E\begin{pmatrix}
\psi_0 \\
\psi_1 \\
\vdots \\
\psi_{N-2} \\
\psi_{N-1}
\end{pmatrix}
$$

In this manner, we have mapped our differential equation onto a matrix eigenvalue problem, $H\psi = E\psi$, where $H$ is the sum of the two matrices in the equation above (including their constant prefactors). By diagonalizing this matrix numerically, we obtain the eigenvalues and eigenvectors that corresponding to the system's allowed energies and wavefunctions, respectively. There are a number of different libraries that can perform these diagonalizations efficiently, especially since the matrix in question is sparse, with only the upper-right/lower-left elements keeping it from being tridiagonal. 
 

```python
# Imports, matrix functions, and diagonalization routine

import numpy as np
import matplotlib.pyplot as plt
from scipy import sparse
from scipy.sparse import linalg as sla
from scipy.special import mathieu_cem, mathieu_sem, mathieu_a, mathieu_b # Mathieu functions and characteristic values

from tqdm.notebook import tqdm
%pip install ipywidgets
from IPython.display import display
from ipywidgets import interact, FloatSlider

def V(phi, Ex, R = 1, e = 1):
    """ Perturbing potential introduced by constant electric field."""
    return e*Ex*R*np.cos(phi)

def matrix_H(Ex, N, hbar=1, m=1, R=1, e=1):
    """
    Return an N x N matrix representation of the Hamiltonian for 
    an electric field strength Ex. 
    """

    phi = np.linspace(0, 2*np.pi, N, endpoint=False)
    dphi = 2*np.pi / N
    
    # Kinetic energy
    t = hbar**2 / (2 * m * R**2 * dphi**2) # Prefix
    ones = np.ones(N)
    T = sparse.diags([-ones[:-1], 2*ones, -ones[:-1]], [-1, 0, 1], format='lil') # Main diagonals
    T[0, N-1] = T[N-1, 0] = -1                                                   # Periodic corners

    return (t * T + sparse.diags(V(phi, Ex))).tocsc()


def diagonalize_H(Ex, N, neigs = 5, hbar=1, m=1, R=1, e=1):
    
    """
    Determines the lowest energy solutions of the perturbed
    particle-on-a-ring system by discretizing the wavefunction
    and converting the Schrodinger equation to a matrix 
    equation. 
    """

    phi = np.linspace(0, 2*np.pi, N, endpoint=False)
    
    H = matrix_H(Ex, N)
          
    [eigvals, eigvecs] = sla.eigsh(H, k = neigs, which='SA')
    
    # Normalize wavefunctions
    for i in range(neigs):
        psi = eigvecs[:, i]
        norm = np.sqrt(np.sum(np.abs(psi)**2, axis=0) * (2*np.pi / N)) 
        eigvecs[:, i] = psi/norm

    return eigvals, eigvecs
```

```python
# Determine the energies for a range of electric field values starting from zero

Ex_vals = np.linspace(0, 0.5, 101) # Electric field values
energies = np.zeros((len(Ex_vals), 5)) 

for i, Ex in enumerate(tqdm(Ex_vals)):
    eigvals, eigvecs = diagonalize_H(Ex, 501, neigs=5)
    energies[i, :] = eigvals

E_0  = energies[:, 0]
E_11 = energies[:, 1]
E_12 = energies[:, 2]
E_21 = energies[:, 3]
E_22 = energies[:, 4]
```

```python
# Plot numerical eigenvalues and compare against perturbation theory results

fig, axs = plt.subplots(1, 3, figsize = (9, 4))
fs = 11

for i in range(3):
    axs[i].set_facecolor('ghostwhite')

axs[0].plot(Ex_vals, E_0, color='black', marker='o', alpha=0.2, label = '$E_0$')
axs[0].plot(Ex_vals, -Ex_vals**2, 'k--', alpha=0.5)
axs[0].text(0.15, -0.15, '$E_0^{(2)} = -F^2$')
axs[0].legend()

axs[1].plot(Ex_vals, E_11, color='darkblue', marker='o', alpha=0.1, label = '$E_{1,0}$')
axs[1].plot(Ex_vals, E_12, color='dodgerblue', marker='o', alpha=0.2, label = '$E_{1,1}$')
axs[1].plot(Ex_vals, 0.5+(5/6)*Ex_vals**2, 'k--', alpha=0.5) 
axs[1].plot(Ex_vals, 0.5-(1/6)*Ex_vals**2, 'k--', alpha=0.5)
axs[1].text(0.05, 0.45, '$E_{1,0}^{(2)} = -\\frac{1}{6}F^2$')
axs[1].text(0.05, 0.60, '$E_{1,1}^{(2)} = +\\frac{5}{6}F^2$')
axs[1].legend()

axs[2].plot(Ex_vals, E_21, color='darkgreen', marker='o', alpha=0.2, label = '$E_{2,0}$')
axs[2].plot(Ex_vals, E_22, color='mediumseagreen', marker='o', alpha=0.2, label = '$E_{2,1}$')
axs[2].plot(Ex_vals, 2+(1/15)*Ex_vals**2, 'k--', alpha=0.5)
axs[2].text(0.05, 2.02, '$E_{2,0}^{(2)} = E_{2,1}^{(2)} = +\\frac{1}{15}F^2$')
axs[2].legend()

for k in range(3):
    axs[k].set_xlabel('Field Strength',fontsize=fs)
    axs[k].set_ylabel('Energy', fontsize=fs)

axs[0].set_title('$n=0$', fontsize=fs+2)
axs[1].set_title('$n=1$', fontsize=fs+2)
axs[2].set_title('$n=2$', fontsize=fs+2)

axs[0].set_ylim(-0.25, 0.05)
axs[1].set_ylim(0.4, 0.7)
axs[2].set_ylim(1.98, 2.04)

axs[0].set_yticks([-0.25, -0.2, -0.15, -0.1, -0.05, 0, 0.05])
axs[1].set_yticks([0.40, 0.45, 0.50, 0.55, 0.60, 0.65, 0.70])
axs[2].set_yticks([1.98, 1.99, 2.00, 2.01, 2.02, 2.03, 2.04])

#axs[0].text(-0.025, 0.06, '(a)' ,fontsize=fs)
#axs[1].text(-0.025, 0.71, '(b)' ,fontsize=fs)
#axs[2].text(-0.025, 2.042, '(c)' ,fontsize=fs)

for k in range(3):
    axs[k].set_xticks([0, 0.15, 0.30, 0.45])

plt.tight_layout()

plt.show()
```

The above plot compares the 'exact' energies obtained by diagonalizing our matrices to our earlier perturbation theory results. For ease of notation, the physical constants $\hbar, m, e$ and $R$ have all been set equal to 1.

In the widget below, we plot the the probability density of the five lowest energy eigenstates. At zero field, these correspond to the 'good states' of the $n=0$, $n=1$, and $n=2$ energy levels, the probability densities of which are shown as faint dotted lines. 

By adjusting the slider, you can explore the effect that different electric field strengths have on the eigenstates. 

We have made no attempt to make the eigenvector routine choose a particular phase for its solutions, or to make it choose the 'good states' out of the infinite number of degenerate eigenvectors when the electric field is zero. While the former ceases to be an issue when we take the modulus squared of our solutions, the effect of the latter can sometimes be seen when dragging the slider through zero. In general we obtain a shifted version of the 'good states', since without an external field, the rotational symmetry of the problem is not broken. 

```python
# Interactive plot of numerical wavefunction probability densities

def modulus_squared(psi):
    """ Return the modulus squared (i.e., the probability density) of a 1D wavefunction"""
    return np.real(np.conj(psi)*psi) # |ψ(x)|^2 = ψ^*(x) ψ(x)

def plot_wavefunctions(Ex):

    N = 101
    eigvals, eigvecs = diagonalize_H(Ex, N)
    phi = np.linspace(0, 2*np.pi*(N-1)/N, N)
    
    plt.cla()
    fig.set_facecolor('ghostwhite')

    offsets = [0, 1, 2] # Vertical offsets for wavefunction plots

    fs = 15              # Font size for axis labels
    transparency = 0.25  # Transparency of 'good state' wavefunction plots
    
    # Ground state
    plt.plot(phi, modulus_squared(eigvecs[:, 0]) +offsets[0], 'k')
    plt.axhline(offsets[0], color='k', alpha=0.25)
    plt.plot(phi, np.ones_like(phi)/(2*np.pi) +offsets[0], color='k', linestyle = '--', alpha = transparency)   # "Good" state

    # First degenerate subspace
    plt.plot(phi, modulus_squared(eigvecs[:, 1])+offsets[1], 'darkblue')
    plt.axhline(offsets[1], color='k', alpha=0.25)
    plt.plot(phi, np.sin(phi)**2/(np.pi) + offsets[1], color='darkblue', linestyle = '--', alpha = transparency)   # "Good" state

    plt.plot(phi, modulus_squared(eigvecs[:, 2])+offsets[1], 'dodgerblue')
    plt.axhline(offsets[1], color='k',alpha=0.25)
    plt.plot(phi, np.cos(phi)**2/(np.pi) + offsets[1], color='dodgerblue', linestyle = '--', alpha = transparency)   # "Good" state

    # Second degenerate subspace
    plt.plot(phi, modulus_squared(eigvecs[:, 3])+offsets[2], 'darkgreen')
    plt.axhline(offsets[2], color='k', alpha=0.25)
    plt.plot(phi, np.sin(2*phi)**2/(np.pi) + offsets[2], color='darkgreen', linestyle = '--', alpha = transparency)   # "Good" state

    plt.plot(phi, modulus_squared(eigvecs[:, 4])+offsets[2], 'mediumseagreen')
    plt.axhline(offsets[2], color='k', alpha=0.25)
    plt.plot(phi, np.cos(2*phi)**2/(np.pi) + offsets[2], color='mediumseagreen', linestyle = '--', alpha = transparency)   # "Good" state
    
    plt.xlabel(r'$\phi$', fontsize=fs)
    plt.ylabel(r'$ |\psi_{n,i}|^2$', fontsize=fs)
    plt.ylim(-0.05, 2.55)
    plt.yticks(offsets, [])

plt.figure()
interact(plot_wavefunctions, Ex=FloatSlider(min=-5, max=20, step=0.02, value=0.01))
plt.show()
```

# Analytic Solutions (Mathieu Functions)

An alternative approach to obtaining the exact eigenvalues/eigenfunctions of our system exploits the fact that our Schrodinger equation is equivalent to the **Mathieu differential equation**, a well-known problem whose solutions are available in a number of different scientific programming libraries, including SciPy. 

The angular Schrodinger equation for this problem is given by 

$$ -\frac{\hbar^2}{2mR^2} \frac{d^2 \psi}{d\phi^2} +eFR \cos(\phi) \psi = E\psi, $$

where $E$ is the energy eigenvalue. 

If we define the rotational energy scale $B = \frac{\hbar^2}{2mR^2}$, this can be re-written as

$$ \frac{d^2 \psi}{d\phi^2} + (\frac{E}{B} - \frac{eFR}{B} \cos \phi)\psi = 0 $$

If we define $x = \frac{\phi}{2}, a = \frac{4E}{B},$ and $q = \frac{2eFR}{B}$, the differential equation can be re-written in the following form:

$$ \frac{d^2 \psi}{dx^2} + \bigg( a - 2q \cos (2x) \bigg) \psi = 0, $$

which is the standard form of the Mathieu differential equation. 

Our solutions must be periodic functions of $\phi$ with period $2\pi$, or equivalently, periodic functions of $x$ with period $\pi$. Note that this is not possible for all values of $a$. For a given value of $q$, there are infinitely many values of $a$ that produce solutions with period $\pi$, but these values of $a$ are **discrete**. These so-called **characteristic values** are usually broken up into two sequences: $a_n(q)$ and $b_n(q)$. The former correspond to even eigenfunctions, while the latter correspond to odd eigenfunctions. 

The standard even/odd solutions of this second-order differential equation - known as *Mathieu functions* - are written as 

$$ ce_n(x, q), \quad cs_n(x, q) $$

For integer $n$, these functions have period $\pi$ when $n$ is even, and period $2\pi$ when $n$ is odd. Our eigenfunctions are therefore the Mathieu functions with even indices. In other words, we can express our exact (but not yet normalized) eigenfunctions as

$$ \psi_{n,0}(\phi) = ce_{2n}(\phi/2, q), \quad \psi_{n,1}(\phi) = se_{2n}(\phi/2, q), \quad n \in 0, 1, 2, 3, ...$$

with the caveat that for the ground state ($n=0$), we ignore the second solution. 

Note that in the limit of a vanishing electric field, $q\to 0$, and the Mathieu functions reduce to the following trigonometric functions:

$$ ce_{2n}(\phi/2, q\to 0) = \cos(n \phi) $$

$$ se_{2n}(\phi/2, q\to 0) = \sin(n \phi) $$

In other words, the 'good states' for degenerate perturbation theory calculations are simple trigonometric functions. This generalizes our earlier result for the $n=1$ degenerate subspace. 

The energies for the even states are given by 

$$ E = \frac{B}{4} a_{2n}(q) = \frac{\hbar^2}{8mR^2} a_{2n}(q), \quad n \in 0, 1, 2, 3, ...$$ 

and for odd states by 

$$ E = \frac{B}{4} b_{2n}(q) = \frac{\hbar^2}{8mR^2} b_{2n}(q), \quad n \in 1, 2, 3,...$$ 

```python
# Define Matheiu function eigenstates
def psi(n, i, q, phi):
    """
    Return the i-th normalized wavefunction in the n-th
    degenerate subspace of the unperturbed ring.

    n = 0: ground-state branch; i = 0
    n > 0: i = 0 (even), i = 1 (odd)
    """
    x_deg = np.degrees(phi / 2)

    if n == 0:
        if i != 0:
            raise ValueError("For n=0, i must be 0.")
        y, _ = mathieu_cem(0, q, x_deg)
    elif n > 0:
        if i == 0:
            y, _ = mathieu_cem(2*n, q, x_deg)
        elif i == 1:
            y, _ = mathieu_sem(2*n, q, x_deg)
        else:
            raise ValueError("For n > 0, i must be 0 or 1.")
    else:
        raise ValueError("n must be nonnegative.")

    norm = np.sqrt(np.trapezoid(y**2, phi)) # Applies half-weight to boundaries, which should not happen for a periodic system
    return y / norm
```

```python
# Plot Mathieu function probability densities

N = 100 
phi = np.linspace(0, 2*np.pi*(N-1)/N, N)

F = 1
q = 4*F

n = 1

plt.figure()

offsets = [0, 1, 2]
for offset in offsets:
    plt.axhline(offset, color = 'k', alpha = 0.5)

n=0
plt.plot(phi, psi(n, 0, q, phi)**2 + offsets[n], color = 'black')

n=1
plt.plot(phi, psi(n, 0, q, phi)**2 + offsets[n], color = 'dodgerblue')
plt.plot(phi, psi(n, 1, q, phi)**2 + offsets[n], color = 'darkblue')

n=2
plt.plot(phi, psi(n, 0, q, phi)**2 + offsets[n], color = 'mediumseagreen')
plt.plot(phi, psi(n, 1, q, phi)**2 + offsets[n], color = 'darkgreen')

plt.show()
```
