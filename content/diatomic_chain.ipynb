# Diatomic Chain

We now explore how the properties of the phonon modes change if the atomic chain contains two different types of atoms, i.e. the properties of the so-called **diatomic chain**. Technically we are going to explore how the properties change if the chain has two different *spring constants*, because the algebra is a little easier, but the resulting physics is almost identical. 

$$ m \ddot{\delta x}_n = \kappa_2(\delta y_n -\delta x_n) + \kappa_1(\delta y_{n-1} - \delta x_n) $$

$$ m \ddot{\delta y}_n = \kappa_1(\delta x_{n+1} -\delta y_n) + \kappa_2(\delta x_n - \delta y_n) $$

Just like in the monoatomic case, we propose as an ansatze that these quantities should be wave-like:

$$ \delta x_n = A_x e^{i\omega t - ikna} $$

$$ \delta y_n = A_y e^{i\omega t - ikna} $$

Plugging our ansatze into our equations of motion, we find that 

$$ -\omega^2 m A_x e^{i\omega t - ikna} = \kappa_2 A_y e^{i \omega t - ikna} + \kappa_1 A_y e^{i\omega t - ik(n-1)a} - (\kappa_1 + \kappa_2)A_x e^{i\omega t - ikna} $$

$$
-\omega^2 m A_y e^{i\omega t-ikna}
=
\kappa_1 A_x e^{i\omega t-ik(n+1)a}
+\kappa_2 A_x e^{i\omega t-ikna}
-(\kappa_1+\kappa_2)A_y e^{i\omega t-ikna}.
$$

which simplifies to

$$
-\omega^2 m A_x
=
\kappa_2 A_y
+\kappa_1 A_y e^{ika}
-(\kappa_1+\kappa_2)A_x
$$

$$
-\omega^2 m A_y
=
\kappa_1 A_x e^{-ika}
+\kappa_2 A_x
-(\kappa_1+\kappa_2)A_y.
$$

This can be rewritten conveniently as an eigenvalue equation:

$$
m\omega^2
\begin{pmatrix}
A_x\\
A_y
\end{pmatrix}
=
\begin{pmatrix}
(\kappa_1+\kappa_2) & -\kappa_2-\kappa_1e^{ika}\\
-\kappa_2-\kappa_1e^{-ika} & (\kappa_1+\kappa_2)
\end{pmatrix}
\begin{pmatrix}
A_x\\
A_y
\end{pmatrix}.
$$

The solutions of this are obtained by finding the zeros of the characteristic determinant:

$$
0=
\begin{vmatrix}
(\kappa_1+\kappa_2)-m\omega^2
&
-\kappa_2-\kappa_1e^{ika}
\\
-\kappa_2-\kappa_1e^{-ika}
&
(\kappa_1+\kappa_2)-m\omega^2
\end{vmatrix}
$$

$$
=
\left|(\kappa_1+\kappa_2)-m\omega^2\right|^2
-
\left|\kappa_2+\kappa_1e^{ika}\right|^2.
$$

the roots of which are clearly given by

$$
m\omega^2
=
(\kappa_1+\kappa_2)
\pm
\left|\kappa_1+\kappa_2e^{ika}\right|.
$$

The second term here can be simplified to

$$
\left|\kappa_1+\kappa_2e^{ika}\right|
=
\sqrt{
(\kappa_1+\kappa_2e^{ika})
(\kappa_1+\kappa_2e^{-ika})
}
$$

$$
=
\sqrt{
\kappa_1^2+\kappa_2^2
+2\kappa_1\kappa_2\cos(ka)
}.
$$

so we finally obtain

$$
\omega_{\pm}
=
\sqrt{
\frac{\kappa_1+\kappa_2}{m}
\pm
\frac{1}{m}
\sqrt{
\kappa_1^2+\kappa_2^2
+2\kappa_1\kappa_2\cos(ka)
}
}.
$$

$$
=
\sqrt{
\frac{\kappa_1+\kappa_2}{m}
\pm
\frac{1}{m}
\sqrt{
(\kappa_1+\kappa_2)^2
-4\kappa_1\kappa_2\sin^2(ka/2)
}
}.
$$

```python
def diatomic_omega_optical(k, kappa1 = 1, kappa2 = 1, m = 1, a = 1):
    """
    Returns the frequency of the optical mode of the diatomic chain with wavevector k. 
    """
    w = np.sqrt( (kappa1+kappa2)/m + (1/m)*np.sqrt((kappa1+kappa2)**2 -4*kappa1*kappa2*np.sin(k*a/2)**2) )
    
    return w

def diatomic_omega_acoustic(k, kappa1 = 1, kappa2 = 1, m = 1, a = 1):
    """
    Returns the frequency of the acoustic mode of the diatomic chain with wavevector k. 
    """
    w = np.sqrt( (kappa1+kappa2)/m - (1/m)*np.sqrt((kappa1+kappa2)**2 -4*kappa1*kappa2*np.sin(k*a/2)**2) )
    
    return w
```
