```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

def plot_frequency(frequency=2):
    x = np.linspace(0, 2 * np.pi, 500)
    y = np.sin(frequency * x)

    plt.figure(figsize=(8, 4))
    plt.plot(x, y)
    plt.ylim(-1.1, 1.1)
    plt.grid()
    plt.show()

widgets.interact(
    plot_frequency,
    frequency=widgets.FloatSlider(
        value=2,
        min=0.5,
        max=10,
        step=0.5
    )
);
```
